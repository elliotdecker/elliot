```python
# 3 hr_diagram
# 3.1 Importing Data with NumPy
import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt("../../course_assignments/homework7/star_data.txt",
skiprows=1)
```

```python
# 3.2 Analyzing Imported Data
print(data.shape)
# there are 2642 rows

magnitude = data[:, 0]
temperature = data[:, 1]

print(magnitude.min())
print(magnitude.max())
print(magnitude.argmin())
print(magnitude.argmax())
# star #1988 has the lowest magnitude and star #264 has the greatest

print(temperature.min())
print(temperature.max())
print(temperature.argmin())
print(temperature.argmax())
# star #1923 has the lowest temperature and star #2613 has the greatest

# star # 1988 is the brightest and star #264 is the dimmest
```

```python
# 3.3 Plot HR Diagram
brightest = magnitude.argmin()
dimmest = magnitude.argmax()
hottest = temperature.argmax()
coldest = temperature.argmin()

log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)

plt.figure(figsize=(8, 6))

plt.scatter(log_temperature, log_magnitude, color="blue", s=30, alpha=0.6, label="Stars")

plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

plt.scatter(log_temperature[brightest], log_magnitude[brightest],
            color="black", marker="*", s=300, label="Brightest Star")
plt.scatter(log_temperature[dimmest],  log_magnitude[dimmest],
            color="purple", marker="*", s=300, label="Dimmest Star")
plt.scatter(log_temperature[hottest],  log_magnitude[hottest],
            color="green", marker="*", s=300, label="Hottest Star")
plt.scatter(log_temperature[coldest],  log_magnitude[coldest],
            color="orange", marker="*", s=300, label="Coldest Star")

plt.title("Hertzsprung-Russell Diagram")

plt.xlabel("Temperature (log K)")

plt.ylabel("Magnitude")

plt.legend()

plt.show()
```
