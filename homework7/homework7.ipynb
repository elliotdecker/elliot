```python
# Homework 7
# 2 Matplotlib
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sine_wave(x, A, w):
    return A * np.sin(w * x)

x = np.linspace(0, 2*np.pi, 1000)

amplitude = np.array([0.5, 1, 1.5, 2, 2.5])
frequency = np.array([1, 2, 3, 4, 5])

fig, ax = plt.subplots(1, 3, figsize=(25, 5))

for A in amplitude:
    y = make_sine_wave(x, A, w=1)
    ax[0].plot(x, y, label=f"A = {A}")

for w in frequency:
    y = make_sine_wave(x, A=1, w=w)
    ax[1].plot(x, y, label=f"w = {w}")

for A, w in zip(amplitude, frequency):
    y = make_sine_wave(x, A, w)
    ax[2].plot(x, y, label=f"A={A}, w={w}")

ax[0].set_title("Varying Amplitude (ω = 1)", fontsize=15)
ax[0].set_xlabel("x", fontsize=12)
ax[0].set_ylabel("y", fontsize=12)
ax[0].legend()

ax[1].set_title("Varying Frequency (A = 1)", fontsize=15)
ax[1].set_xlabel("x", fontsize=12)
ax[1].set_ylabel("y", fontsize=12)
ax[1].legend()

ax[2].set_title("Varying Amplitude and Frequency Together", fontsize=15)
ax[2].set_xlabel("x", fontsize=12)
ax[2].set_ylabel("y", fontsize=12)
ax[2].legend()

plt.show()
```
